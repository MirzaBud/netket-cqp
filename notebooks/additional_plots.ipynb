## plotting models for the various d/h vals

Relative error and computation time versus degrees of freedom.


```python
import re
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

%config InlineBackend.figure_format = "retina"

plt.rcParams.update({
    "text.usetex": False,
    "font.family": "serif",
    "mathtext.fontset": "cm",
    "figure.dpi": 180,
    "savefig.dpi": 600,
    "axes.linewidth": 1.2,
    "font.size": 15,
    "axes.labelsize": 18,
    "axes.titlesize": 16,
    "legend.fontsize": 14,
    "xtick.labelsize": 15,
    "ytick.labelsize": 15,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
})

def comp_time_to_minutes(value):
    text = str(value).strip()
    match = re.fullmatch(r"(?:(\d+)h\s*)?(?:(\d+)m\s*)?(?:(\d+)s)?", text)
    if match is None:
        raise ValueError(f"Could not parse computation time: {value!r}")

    hours, minutes, seconds = match.groups(default="0")
    return int(hours) * 60 + int(minutes) + int(seconds) / 60

plot_df = dataframes.copy().sort_values("DOF")
plot_df["comp_time_min"] = plot_df["comp_time"].map(comp_time_to_minutes)
plot_df["rel_error_plot"] = abs(plot_df["E_vit"] - e_exact) / abs(e_exact)

model_labels = [
    fr"$(d,h)=({int(row.d)},{int(row.h)})$"
    for _, row in plot_df.iterrows()
]
model_box = "Models: " + ", ".join(model_labels[:3])
if len(model_labels) > 3:
    model_box += "\n" + " " * 8 + ", ".join(model_labels[3:])

blue = "#2171b5"
green = "#31a354"

fig, axes = plt.subplots(1, 2, figsize=(10.5, 4.2))

axes[0].plot(
    plot_df["DOF"],
    plot_df["rel_error_plot"],
    color=blue,
    marker="s",
    markersize=7,
    markeredgecolor="black",
    markeredgewidth=0.9,
    linewidth=1.5,
    label=r"$J_2/J_1 = 0.4$",
)
axes[0].set_xscale("log")
axes[0].set_yscale("log")
axes[0].set_xlabel("Degrees of freedom")
axes[0].set_ylabel("Relative error")
axes[0].set_title("Relative error")
axes[0].legend(frameon=True, fancybox=True, borderpad=0.5)

axes[1].plot(
    plot_df["DOF"],
    plot_df["comp_time_min"],
    color=green,
    marker="o",
    markersize=7,
    markeredgecolor="black",
    markeredgewidth=0.9,
    linewidth=1.5,
    label=r"$J_2/J_1 = 0.4$",
)
axes[1].set_xscale("log")
axes[1].set_yscale("log")
axes[1].set_xlabel("Degrees of freedom")
axes[1].set_ylabel("Computation time [min]")
axes[1].set_title("Computation time")
axes[1].legend(frameon=True, fancybox=True, borderpad=0.5)

for ax in axes:
    ax.tick_params(which="both", direction="in", top=True, right=True)
    ax.xaxis.set_major_formatter(ticker.ScalarFormatter())
    ax.xaxis.set_minor_locator(ticker.LogLocator(base=10, subs=range(2, 10)))
    ax.yaxis.set_minor_locator(ticker.LogLocator(base=10, subs=range(2, 10)))
    ax.grid(False)

fig.text(
    0.5,
    0.015,
    model_box,
    ha="center",
    va="bottom",
    fontsize=13,
    bbox=dict(boxstyle="round,pad=0.35", facecolor="white", edgecolor="0.8"),
)
fig.tight_layout(rect=(0, 0.14, 1, 1))
plt.savefig("additional_relative_error_and_time_dof.pdf", bbox_inches="tight")
plt.savefig("additional_relative_error_and_time_dof.png", dpi=600, bbox_inches="tight")
plt.show()
```
